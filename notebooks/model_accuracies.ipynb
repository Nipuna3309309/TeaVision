This interactive notebook tracks the performance metrics for the **YOLOv8** object detection models and the **OpenCV + Classical ML** grading models utilized in the Tea Analysis System.

```python
import pandas as pd
import matplotlib.pyplot as plt

# YOLOv8 Models (mAP50 Scores)
yolo_data = {
    'Detection Model': ['TeaNet Roboflow v4', 'TeaNet Plus (Estimated)', 'TeaNet Pro (Estimated)', 'TeaNet Micro (Estimated)', 'TeaNet V2'],
    'mAP50 Score': [0.734, 0.710, 0.680, 0.650, 0.491],
    'Focus': ['Ultimate Accuracy', 'Augmented Range', 'Damage Optimization', 'Small Object Specialty', 'Legacy Baseline']
}
yolo_df = pd.DataFrame(yolo_data)
display(yolo_df.style.set_caption('Tea Detection - YOLOv8 Models').background_gradient(subset=['mAP50 Score'], cmap='Greens'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Classical ML & DL Models (Extracting exact data from backend)
ml_data = [
    {'Model': 'MLP Neural Network', 'Train Accuracy (%)': 100.0, 'Test Accuracy (%)': 86.67, 'F1 Score (%)': 86.85},
    {'Model': 'AdaBoost', 'Train Accuracy (%)': 94.92, 'Test Accuracy (%)': 83.33, 'F1 Score (%)': 83.47},
    {'Model': 'Logistic Regression', 'Train Accuracy (%)': 94.07, 'Test Accuracy (%)': 83.33, 'F1 Score (%)': 83.62},
    {'Model': 'SVM (RBF)', 'Train Accuracy (%)': 94.92, 'Test Accuracy (%)': 83.33, 'F1 Score (%)': 83.47},
    {'Model': 'SVM (Linear)', 'Train Accuracy (%)': 95.76, 'Test Accuracy (%)': 83.33, 'F1 Score (%)': 83.62},
    {'Model': 'Naive Bayes', 'Train Accuracy (%)': 90.68, 'Test Accuracy (%)': 83.33, 'F1 Score (%)': 83.62},
    {'Model': 'Random Forest', 'Train Accuracy (%)': 94.07, 'Test Accuracy (%)': 80.0, 'F1 Score (%)': 80.0},
    {'Model': 'Gradient Boosting', 'Train Accuracy (%)': 99.15, 'Test Accuracy (%)': 76.67, 'F1 Score (%)': 75.68},
    {'Model': 'Decision Tree', 'Train Accuracy (%)': 95.76, 'Test Accuracy (%)': 76.67, 'F1 Score (%)': 76.86},
    {'Model': 'KNN', 'Train Accuracy (%)': 90.68, 'Test Accuracy (%)': 76.67, 'F1 Score (%)': 77.06},
]
ml_df = pd.DataFrame(ml_data).sort_values(by='Test Accuracy (%)', ascending=False).reset_index(drop=True)
display(ml_df.style.set_caption('Grade My Tea - Classification Models').format(precision=2))
```

```python
# Visualizing the ML Models Performance
plt.figure(figsize=(14, 6))
x = range(len(ml_df))
bars = plt.bar(x, ml_df['Test Accuracy (%)'], color='#3F51B5')
plt.xticks(x, ml_df['Model'], rotation=45, ha='right', fontsize=11)
plt.ylabel('Test Accuracy (%)', fontsize=12)
plt.title('ML Classification Models - Test Accuracy Comparison', fontsize=14, fontweight='bold')
plt.ylim(0, 105)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1.5, f'{yval}%', ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.show()
```
